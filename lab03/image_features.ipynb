# Part B: Image Features
Load the images from the Cats and Dogs folders, resize them, convert to grayscale, and build a feature matrix.

```python
from pathlib import Path
import sys
import numpy as np

candidates = [Path.cwd(), Path.cwd() / 'Lab 3', Path.cwd().parent, Path.cwd().parent / 'Lab 3']
LAB_DIR = next((p for p in candidates if (p / 'Cats').is_dir() and (p / 'Dogs').is_dir()), None)
if LAB_DIR is None:
    raise FileNotFoundError('Could not locate the Cats and Dogs folders.')
sys.path.insert(0, str(LAB_DIR / 'lab03'))
from image_utils import image_to_features

IMAGE_EXTENSIONS = {'.jpg', '.jpeg', '.png'}
SIZE = (64, 64)
X_rows, y_rows = [], []
for label, folder_name in enumerate(('Cats', 'Dogs')):
    folder = LAB_DIR / folder_name
    paths = sorted(p for p in folder.iterdir() if p.suffix.lower() in IMAGE_EXTENSIONS)
    if len(paths) != 10:
        raise ValueError(f'Expected 10 images in {folder}, found {len(paths)}')
    for path in paths:
        pixels = image_to_features(path, size=SIZE).astype(np.float32) / 255.0
        X_rows.append(pixels)
        y_rows.append(label)

X = np.stack(X_rows)
y = np.asarray(y_rows, dtype=np.int64)
dataset_path = LAB_DIR / 'lab03' / 'dataset.npz'
np.savez(dataset_path, X=X, y=y)

print('Class labels: 0 = Cats, 1 = Dogs')
print('Samples:', len(y))
print('Features per sample:', X.shape[1])
print('Feature matrix shape:', X.shape)
print('Label vector shape:', y.shape)
print('Saved:', dataset_path)
```
